  1|   The Mutable Default Trap: Write a function called add_item(item, basket=[]) that appends an item to the basket and returns it. Call the function twice without passing a basket argument and print the results to observe why mutable defaults break.

```python
def add_item(item,basket=[]):
  basket.append(item)
  return (basket)

add_item("apple")
print(add_item("banana"))
```

The main thing is that is the none value is assigned then it is optional to take value as parameter, like in basket=[]
