Let's break down what happens in each zone:

Before / (Positional-Only): Must be passed strictly by position.

Between / and *args (Standard): Can be passed by position or by keyword.

*args (Positional Collector): Packs any extra leftover positional arguments into a tuple.

After * or *args (Keyword-Only): Must be passed explicitly by keyword name. (Note: Any parameter placed after *args automatically becomes keyword-only, but you can also use a standalone * if you don't have *args).

**kwargs (Keyword Collector): Packs any extra leftover keyword arguments into a dictionary. (This must always go at the very end).

```python
def process_order(store_id, /, item_name, *extras, discount=0.0, express_shipping, **notes):
    print(f"Store (Positional-only): {store_id}")
    print(f"Item (Standard): {item_name}")
    print(f"Extras (*args tuple): {extras}")
    print(f"Discount (Keyword with default): {discount}")
    print(f"Express Shipping (Keyword-only): {express_shipping}")
    print(f"Notes (**kwargs dict): {notes}")

# --- Valid Call ---
process_order(
    101,                  # store_id (must be positional, left of /)
    "Laptop",             # item_name (can be positional)
    "Mouse", "Keyboard",  # *extras (extra positional arguments caught into a tuple)
    discount=0.15,        # discount (passed by keyword)
    express_shipping=True,# express_shipping (keyword-only, right of *args)
    gift_wrap=True,       # notes (**kwargs, extra keyword arguments caught into a dict)
    comment="Leave at front door"
)
```

```python
def make_request(url,/,method="GET",*args,timeout=30,retries=3,**kwargs):
  return{
    "url":"/".join((url,*args)),
    "method":method,
    "timeout": timeout,
    "retries": retries,
    "headers": kwargs
  }

print(make_request("https://api.com"))
print(make_request("https://api.com", "POST", "users", "42", timeout=10, Auth="token"))
print(make_request("https://api.com", "PUT", retries=5))
```

```python
def calculate_discount(price, discount_percent):
    """Calculate the final price after applying a percentage discount.

    Args:
        price (float): The original price. Must not be negative.
        discount_percent (float): The discount percentage, from 0 to 100.

    Returns:
        float: The final price after the discount, rounded to 2 decimal places.

    Raises:
        ValueError: If price is negative, or if discount_percent is
            below 0 or above 100.
    """
    if price < 0 or discount_percent < 0 or discount_percent > 100:
        raise ValueError("price must be >= 0 and discount_percent must be 0-100")

    discount_amount = price * discount_percent / 100
    return round(price - discount_amount, 2)


print(calculate_discount(200, 10))
print(calculate_discount(99.99, 15))
print(calculate_discount(50, 0))
help(calculate_discount)
```
