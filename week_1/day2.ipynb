2. Prime checker, then list every prime below 100.

```python
n=int(input("enter number"))
def is_prime(n):
    if n <= 1:
        return False

    # Check divisors from 2 up to the square root of n
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False  # Found a divisor, so it's not prime

    return True  # No divisors found, it's prime

is_prime(n)
```

☐ 3. Number-guessing game with while and higher/lower hints (limit 7 tries)

```python
from random import choice
import random
limit =0
numbers=[1,2,3,4,5,6,8,7,9,0]
guess_number=random.choice(numbers)
while limit <=7:
  number=int(input("Enter a number :"))
  if number < guess_number:
    print("number is less")

  elif number == guess_number:
    print("number is correct")
    break

  else:
    print("number is big")
  limit+=1
```

 4. Print a formatted multiplication table (1-10) with nested loops.

```python
for i in range(1,11):
  print("------------------------------------")
  for j in range(1,11):
    print(f" {i} x {j} = {i*j}")
  print("\n")
```

☐ 5. Pair names and scores with zip, print a ranked list using enumerate(start=1).

```python
while True:
  try:
    n=int(input(" enter how many poeple you want : "))
    num_list=[]
    name_list=[]
    for i in range(n):
      stdn_name=input("enter name : ")
      stdn_num=int(input("enter number : "))
      name_list.append(stdn_name)
      num_list.append(stdn_num)
    final_detail=zip(name_list,num_list)
    for name,score in enumerate(final_detail,start=1):
      print(name,score)
    break

  except ValueError:
    print("type correct datatype ")
```
